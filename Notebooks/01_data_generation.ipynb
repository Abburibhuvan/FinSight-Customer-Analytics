```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customer_id = [
    "CUST" + str(i).zfill(5)
    for i in range(1, 50001)
]
age = np.random.normal(
    loc=25,
    scale=8,
    size=50000
)
age = np.clip(age, 18, 80)
age = np.round(age).astype(int)
gender = np.random.choice(
    ["Male", "Female", "Other"],
    size=50000,
    p=[0.48, 0.47, 0.05]
)
countries = [
    "India",
    "USA",
    "UK",
    "Germany",
    "Canada",
    "Australia",
    "Singapore",
    "UAE"
]
country = np.random.choice(
    countries,
    size=50000,
    p=[0.35, 0.20, 0.10, 0.08, 0.08, 0.07, 0.06, 0.06]
)
employment_types = [
    "Student",
    "Employed",
    "Self-Employed",
    "Business Owner",
    "Retired",
    "Unemployed"
]
employment_type = np.random.choice(
    employment_types,
    size=50000,
    p=[0.12, 0.48, 0.15, 0.10, 0.08, 0.07]
)
income = np.zeros(50000)

income[employment_type == "Student"] = np.random.lognormal(
    mean=np.log(150000),
    sigma=0.7,
    size=(employment_type == "Student").sum()
)

income[employment_type == "Employed"] = np.random.lognormal(
    mean=np.log(800000),
    sigma=0.6,
    size=(employment_type == "Employed").sum()
)
income[employment_type == "Self-Employed"] = np.random.lognormal(
    mean=np.log(1200000),
    sigma=0.7,
    size=(employment_type == "Self-Employed").sum()
)
income[employment_type == "Business Owner"] = np.random.lognormal(
    mean=np.log(2500000),
    sigma=0.9,
    size=(employment_type == "Business Owner").sum()
)
income[employment_type == "Retired"] = np.random.lognormal(
    mean=np.log(600000),
    sigma=0.6,
    size=(employment_type == "Retired").sum()
)
income[employment_type == "Unemployed"] = np.random.lognormal(
    mean=np.log(150000),
    sigma=0.8,
    size=(employment_type == "Unemployed").sum()
)
income = np.clip(income, 0, 10000000)
income = np.round(income, 2)
account_types = [
    "Savings",
    "Current",
    "Salary",
    "Premium"
]
account_type = np.random.choice(
    account_types,
    size=50000,
    p=[0.55, 0.15, 0.20, 0.10]
)
start_time = pd.Timestamp("2023-01-01")
end_day = pd.Timestamp.today()
signup_days = np.random.randint(
    0,
    (end_day - start_time).days + 1,
    size=50000
)
signup_date = start_time + pd.to_timedelta(
    signup_days,
    unit="D"
)
customer_statuses = [
    "Active",
    "Inactive",
    "Suspended",
    "Closed"
]
customer_status = np.random.choice(
    customer_statuses,
    size=50000,
    p=[0.70, 0.15, 0.05, 0.10]
)
```

```python
customers = pd.DataFrame({
    "customer_id": customer_id,
    "age": age,
    "gender": gender,
    "country": country,
    "employment_type": employment_type,
    "income": income,
    "account_type": account_type,
    "signup_date": signup_date,
    "customer_status": customer_status
})
```
